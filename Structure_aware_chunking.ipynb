-->import re
This code uses Python's re (regex) library to recognize meaningful headings and patterns in the extracted PDF text, which helps your RAG pipeline understand the document's structure instead of blindly splitting it by character count.

```python
import re

for page_number, page in enumerate(pages, start=1):

    print("\n" + "=" * 80)
    print(f"PAGE {page_number}")
    print("=" * 80)

    lines = [
        line.strip()
        for line in page.page_content.splitlines()
        if line.strip()
    ]

    for line in lines:

        # -----------------------------
        # PAGE 1 - DOCUMENT INFORMATION
        # -----------------------------
        if re.search(r"Maharashtra Education Society", line, re.I):
            print("DOCUMENT:", line)

        elif re.search(r"ABASAHEB GARWARE COLLEGE", line, re.I):
            print("DOCUMENT:", line)

        elif re.search(r"Three year B\.\s*S?c\.\s*Degree Program", line, re.I):
            print("DOCUMENT:", line)

        elif re.search(r"S\.\s*Y\.\s*B\.\s*Sc", line, re.I):
            print("DOCUMENT:", line)

        elif re.search(r"Academic Year", line, re.I):
            print("DOCUMENT:", line)

        # -----------------------------
        # PAGE 2 - COURSE INFORMATION
        # -----------------------------
        elif re.search(r"TITLE OF THE COURSE", line, re.I):
            print("COURSE INFORMATION:", line)

        elif re.search(r"ELIGIBILITY", line, re.I):
            print("COURSE INFORMATION:", line)

        elif re.search(r"STRUCTURE OF THE COURSE", line, re.I):
            print("COURSE STRUCTURE:", line)

        # -----------------------------
        # SEMESTER
        # -----------------------------
        elif re.search(r"Semester\s*[–-]\s*(III|IV)", line, re.I):
            print("SEMESTER:", line)

        # -----------------------------
        # PAPER
        # -----------------------------
        elif re.search(r"Paper Title", line, re.I):
            print("PAPER:", line)

        # -----------------------------
        # COURSE OUTCOMES
        # -----------------------------
        elif re.search(r"Course outcomes?", line, re.I):
            print("COURSE OUTCOMES")

        # -----------------------------
        # UNITS
        # -----------------------------
        elif re.search(r"Units and Contents", line, re.I):
            print("UNITS")

        elif re.match(r"^Unit\s+\d+", line, re.I):
            print("UNIT:", line)

        # -----------------------------
        # REFERENCES
        # -----------------------------
        elif re.search(r"Reference Books", line, re.I):
            print("REFERENCES")

        # -----------------------------
        # EXPERIMENTS
        # -----------------------------
        elif re.search(r"Expt\.\s*No", line, re.I):
            print("EXPERIMENT TABLE")

        elif re.search(r"Total no\. of experiments", line, re.I):
            print("EXPERIMENT TOTAL")

        # -----------------------------
        # PAPER METADATA
        # -----------------------------
        elif re.search(r"Paper Code", line, re.I):
            print("PAPER METADATA")

        # -----------------------------
        # PROJECT
        # -----------------------------
        elif re.search(
            r"Project Report|Project group|project report",
            line,
            re.I
        ):
            print("PROJECT SECTION")
```

This code extracts each experiment from Practical Paper-I and creates one separate chunk (Document) for each experiment.

```python
#create structure based chunks(its for practical paper I)
experiment_chunks = []

text = pages[9].page_content
lines = text.split("\n")

current_experiment = None
current_text = []

for line in lines:

    line = line.strip()

    if not line:
        continue

    # Case 1: Experiment number and title are on the same line
    match = re.match(r"^(\d{1,2})\s+(.*)", line)

    # Case 2: Experiment number appears alone on a line
    standalone = re.match(r"^(\d{1,2})$", line)

    if match or standalone:

        # Save previous experiment
        if current_experiment is not None:

            experiment_chunks.append(
                Document(
                    page_content=(
                        f"Practical Paper-I. "
                        f"Experiment {current_experiment}: "
                        + " ".join(current_text)
                    ),
                    metadata={
                        "page": pages[9].metadata.get("page", 0) + 1,
                        "type": "experiment",
                        "experiment_number": current_experiment
                    }
                )
            )

        # Start new experiment
        if match:
            current_experiment = match.group(1)
            current_text = [match.group(2)]

        else:
            current_experiment = standalone.group(1)
            current_text = []

    else:

        # Continue adding text to current experiment
        if current_experiment is not None:
            current_text.append(line)


# Save the last experiment
if current_experiment is not None:

    experiment_chunks.append(
        Document(
            page_content=(
                f"Practical Paper-I. "
                f"Experiment {current_experiment}: "
                + " ".join(current_text)
            ),
            metadata={
                "page": pages[9].metadata.get("page", 0) + 1,
                "type": "experiment",
                "experiment_number": current_experiment
            }
        )
    )


print("Number of experiment chunks:", len(experiment_chunks))
```

This is complete structure-aware chunking code for the entire Statistics_syllabus.pdf.
    PDF
 ↓
Understand document structure
 ↓
Document / Course / Unit / Reference /
Experiment / Metadata / Project
 ↓
50 logical chunks

```python
# ================================================================
# STRUCTURE-AWARE CHUNKING FOR THE COMPLETE PDF
# Statistics_syllabus.pdf
# ================================================================
from langchain_core.documents import Document
print("Document import successful!")
from collections import Counter

# ============================================================================
# 1. RESET STRUCTURE CHUNKS
# ============================================================================

structure_chunks = []


# ============================================================================
# 2. HELPER FUNCTIONS
# ============================================================================

def clean_text(text):
    """Clean excessive spaces while preserving useful content."""
    text = text.replace("\x00", " ")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n\s*\n+", "\n\n", text)
    return text.strip()


def add_chunk(page_content, metadata):
    """Add a Document chunk to structure_chunks."""
    structure_chunks.append(
        Document(
            page_content=page_content.strip(),
            metadata=metadata
        )
    )


# ============================================================================
# 3. DOCUMENT INFORMATION - PAGE 1
# ============================================================================

page1_text = pages[0].page_content

add_chunk(
    clean_text(page1_text),
    {
        "type": "document_info",
        "page": 1,
        "source": "Statistics_syllabus.pdf"
    }
)


# ============================================================================
# 4. COURSE STRUCTURE - PAGE 2
# ============================================================================

page2_text = pages[1].page_content

add_chunk(
    clean_text(page2_text),
    {
        "type": "course_structure",
        "page": 2,
        "source": "Statistics_syllabus.pdf"
    }
)


# ============================================================================
# 5. THEORY PAPER INFORMATION
# ============================================================================

theory_papers = [
    {
        "semester": "III",
        "paper": "Probability Distributions - I",
        "start_page": 3,
        "end_page": 5
    },
    {
        "semester": "III",
        "paper": "Probability Distributions - II",
        "start_page": 7,
        "end_page": 8
    },
    {
        "semester": "IV",
        "paper": "Statistical Methods",
        "start_page": 11,
        "end_page": 12
    },
    {
        "semester": "IV",
        "paper": "Sampling Distributions and Exact Tests",
        "start_page": 14,
        "end_page": 15
    }
]


# ============================================================================
# 6. COURSE OUTCOMES
# ============================================================================
# We extract the Course Outcomes section separately for each paper.
# This creates 4 theory course-outcome chunks.
# ============================================================================

def extract_course_outcomes(text):
    """
    Extract text between 'Course Outcomes' and the next major section.
    """

    lines = text.splitlines()

    start_index = None
    end_index = None

    for i, line in enumerate(lines):

        if re.search(r"Course\s+Outcomes", line, re.I):
            start_index = i
            continue

        if start_index is not None:

            # Stop when Units / Contents / Experiment section begins
            if re.search(
                r"Units\s+and\s+Contents|Expt\.\s*No\.|Units\s*&\s*Contents",
                line,
                re.I
            ):
                end_index = i
                break

    if start_index is None:
        return ""

    if end_index is None:
        end_index = len(lines)

    outcome_lines = lines[start_index:end_index]

    return clean_text("\n".join(outcome_lines))


for paper_info in theory_papers:

    start_page = paper_info["start_page"]

    text = pages[start_page - 1].page_content

    outcome_text = extract_course_outcomes(text)

    if outcome_text:

        add_chunk(
            outcome_text,
            {
                "type": "course_outcomes",
                "semester": paper_info["semester"],
                "paper": paper_info["paper"],
                "page": start_page,
                "source": "Statistics_syllabus.pdf"
            }
        )


# ============================================================================
# 7. PRACTICAL PAPER-I COURSE OUTCOMES
# ============================================================================

page10_text = pages[9].page_content

outcome_text_p1 = extract_course_outcomes(page10_text)

if outcome_text_p1:

    add_chunk(
        outcome_text_p1,
        {
            "type": "course_outcomes",
            "semester": "III",
            "paper": "Practical Paper - I",
            "page": 10,
            "source": "Statistics_syllabus.pdf"
        }
    )


# ============================================================================
# 8. PRACTICAL PAPER-II COURSE OUTCOMES
# ============================================================================

page16_text = pages[15].page_content

outcome_text_p2 = extract_course_outcomes(page16_text)

if outcome_text_p2:

    add_chunk(
        outcome_text_p2,
        {
            "type": "course_outcomes",
            "semester": "IV",
            "paper": "Practical Paper - II",
            "page": 16,
            "source": "Statistics_syllabus.pdf"
        }
    )


# ============================================================================
# 9. UNIT PARSER
# ============================================================================
#
# Important:
# The PDF extraction contains malformed text such as:
#
# [1 0 Lectures]
# [05 Lectu res]
# [06 Lectu res]
#
# Therefore the regex below intentionally handles spaces inside the number
# and inside the word "Lectures".
# ============================================================================

def parse_units_for_paper(paper_info):

    semester = paper_info["semester"]
    paper = paper_info["paper"]

    start_page = paper_info["start_page"]
    end_page = paper_info["end_page"]

    current_unit_number = None
    current_unit_title = None
    current_unit_lectures = None
    current_unit_content = []
    current_unit_pages = []

    for page_number in range(start_page, end_page + 1):

        text = pages[page_number - 1].page_content
        lines = text.splitlines()

        for line in lines:

            stripped = line.strip()

            if not stripped:
                continue

            # ----------------------------------------------------------------
            # Detect Unit heading
            # ----------------------------------------------------------------
            #
            # Handles:
            #
            # Unit 1: Something [10 Lectures]
            # Unit 3: Something [1 0 Lectures]
            # Unit 2: Something [05 Lectu res]
            #
            # ----------------------------------------------------------------

            unit_match = re.match(
                r"^Unit\s+(\d+)\s*:\s*(.*?)"
                r"\[\s*([0-9\s]+)\s*Lectu\s*res?\s*\]",
                stripped,
                re.I
            )

            if unit_match:

                # ------------------------------------------------------------
                # Save previous unit
                # ------------------------------------------------------------

                if current_unit_number is not None:

                    content = "\n".join(
                        text
                        for page_number, text in current_unit_content
                    ).strip()

                    add_chunk(
                        content,
                        {
                            "type": "unit",
                            "semester": semester,
                            "paper": paper,
                            "unit_number": current_unit_number,
                            "unit_title": current_unit_title,
                            "lectures": [current_unit_lectures],
                            "start_page": min(current_unit_pages),
                            "end_page": max(current_unit_pages),
                            "source": "Statistics_syllabus.pdf"
                        }
                    )

                # ------------------------------------------------------------
                # Start new unit
                # ------------------------------------------------------------

                current_unit_number = unit_match.group(1)

                current_unit_title = unit_match.group(2).strip()

                current_unit_lectures = re.sub(
                    r"\s+",
                    "",
                    unit_match.group(3)
                )

                current_unit_content = [
                    (page_number, stripped)
                ]

                current_unit_pages = [
                    page_number
                ]

            else:

                # ------------------------------------------------------------
                # Continue current unit
                # ------------------------------------------------------------

                if current_unit_number is not None:

                    current_unit_content.append(
                        (page_number, stripped)
                    )

                    current_unit_pages.append(
                        page_number
                    )

    # ------------------------------------------------------------------------
    # Save final unit
    # ------------------------------------------------------------------------

    if current_unit_number is not None:

        content = "\n".join(
            text
            for page_number, text in current_unit_content
        ).strip()

        add_chunk(
            content,
            {
                "type": "unit",
                "semester": semester,
                "paper": paper,
                "unit_number": current_unit_number,
                "unit_title": current_unit_title,
                "lectures": [current_unit_lectures],
                "start_page": min(current_unit_pages),
                "end_page": max(current_unit_pages),
                "source": "Statistics_syllabus.pdf"
            }
        )


# Parse all four theory papers

for paper_info in theory_papers:
    parse_units_for_paper(paper_info)


# ============================================================================
# 10. REFERENCES
# ============================================================================
#
# One coherent reference chunk for each theory paper.
# ============================================================================

reference_pages = [
    {
        "semester": "III",
        "paper": "Probability Distributions - I",
        "page": 6
    },
    {
        "semester": "III",
        "paper": "Probability Distributions - II",
        "page": 9
    },
    {
        "semester": "IV",
        "paper": "Statistical Methods",
        "page": 13
    },
    {
        "semester": "IV",
        "paper": "Sampling Distributions and Exact Tests",
        "page": 15
    }
]


for ref in reference_pages:

    text = pages[ref["page"] - 1].page_content

    add_chunk(
        clean_text(text),
        {
            "type": "reference",
            "semester": ref["semester"],
            "paper": ref["paper"],
            "page": ref["page"],
            "source": "Statistics_syllabus.pdf"
        }
    )


# ============================================================================
# 11. EXISTING PRACTICAL PAPER-I EXPERIMENTS
# ============================================================================
#
# IMPORTANT:
# Your Practical Paper-I parser is already working correctly.
#
# It produced 11 experiments and Experiment 9 is confirmed.
#
# We reuse those chunks directly.
# We DO NOT parse Practical Paper-I again.
# ============================================================================

print("Adding existing Practical Paper-I chunks...")

print(
    "Practical Paper-I experiments:",
    len(experiment_chunks)
)

for chunk in experiment_chunks:

    # Make a copy of the metadata so that the original chunks are not changed.
    metadata = dict(chunk.metadata)

    # Make sure it is identified as an experiment.
    metadata["type"] = "experiment"

    structure_chunks.append(
        Document(
            page_content=chunk.page_content,
            metadata=metadata
        )
    )


# ============================================================================
# 12. PRACTICAL PAPER-II EXPERIMENTS
# ============================================================================
#
# Page 16 contains:
#
# 1
# 2
# 3
# 4
# 5
# 6
# 7
# 8
# 9 Project
#
# Experiment 6 has a multiline title.
#
# The PDF also says "Total no. of experiments: 12", but only 1-9 are actually
# present in the extracted experiment table. We do NOT invent experiments
# 10, 11 and 12.
# ============================================================================

print("Adding Practical Paper-II chunks...")


practical_ii_experiments = [

    (
        "1",
        "Fitting of lognormal distribution and computation of expected "
        "frequencies. Plot of expected vs. observed frequencies."
    ),

    (
        "2",
        "Test for proportions and construction of confidence interval. "
        "Verification of result using p-value approach."
    ),

    (
        "3",
        "Test for means and construction of confidence interval. "
        "Verification of result using p-value approach."
    ),

    (
        "4",
        "Test based on chi square distribution "
        "(i) Test for independence of attributes "
        "(2 x 2, r x s contingency table) "
        "(ii) Goodness of fit "
        "(iii) McNemar's test."
    ),

    (
        "5",
        "Tests for population variance (one sample and two sample)."
    ),

    (
        "6",
        "Computation of probabilities of continuous uniform, exponential, "
        "normal, log normal, gamma, chi square, t, and F distributions "
        "using R software. Fitting of trivariate regression plane using "
        "R software."
    ),

    (
        "7",
        "Tests using R software for: "
        "(i) Proportions and means. "
        "(ii) Chi square, Student's-t and Snedecor's-F distributions. "
        "(iii) Significance of correlation coefficient and linear "
        "regression coefficient."
    ),

    (
        "8",
        "Model sampling from standard discrete and continuous "
        "distributions using R software."
    ),

    (
        "9",
        "Project: Project based on analysis of data collected by students "
        "in groups of maximum 6 students. "
        "(Project is equivalent to four practicals)"
    )
]


for exp_number, exp_title in practical_ii_experiments:

    add_chunk(
        f"Practical Paper-II. Experiment {exp_number}: {exp_title}",
        {
            "type": "experiment",
            "semester": "IV",
            "paper": "Practical Paper - II",
            "experiment_number": exp_number,
            "start_page": 16,
            "end_page": 16,
            "source": "Statistics_syllabus.pdf"
        }
    )


# ============================================================================
# 13. PRACTICAL PAPER METADATA
# ============================================================================

# Practical Paper-I metadata

add_chunk(
    "Practical Paper-I | Semester III | "
    "Paper Code: USSTP-233 | Number of credits: 2 | "
    "Total no. of experiments: 11 | "
    "Each practical is of duration: 4 hours 20 minutes.",
    {
        "type": "course_metadata",
        "semester": "III",
        "paper": "Practical Paper - I",
        "paper_code": "USSTP-233",
        "credits": 2,
        "page": 10,
        "source": "Statistics_syllabus.pdf"
    }
)


# Practical Paper-II metadata

add_chunk(
    "Practical Paper-II | Semester IV | "
    "Paper Code: USSTP-243 | Number of credits: 2 | "
    "Total no. of experiments: 12 | "
    "Each practical is of duration: 4 hours 20 minutes.",
    {
        "type": "course_metadata",
        "semester": "IV",
        "paper": "Practical Paper - II",
        "paper_code": "USSTP-243",
        "credits": 2,
        "page": 16,
        "source": "Statistics_syllabus.pdf"
    }
)


# ============================================================================
# 14. PROJECT - PAGE 17
# ============================================================================

page17_text = pages[16].page_content

add_chunk(
    clean_text(page17_text),
    {
        "type": "project",
        "page": 17,
        "source": "Statistics_syllabus.pdf"
    }
)


# ============================================================================
# 15. FINAL SUMMARY
# ============================================================================

print("\n")
print("=" * 80)
print("STRUCTURE-AWARE CHUNKING COMPLETE")
print("=" * 80)

print("\nTotal chunks:", len(structure_chunks))

print("\nChunks by type:")

chunk_counts = Counter(
    chunk.metadata.get("type")
    for chunk in structure_chunks
)

for chunk_type, count in chunk_counts.items():
    print(f"{chunk_type:<20}: {count}")


# ============================================================================
# 16. CHECK EXPECTED COUNTS
# ============================================================================

print("\n")
print("=" * 80)
print("EXPECTED STRUCTURE CHECK")
print("=" * 80)

expected_counts = {
    "document_info": 1,
    "course_structure": 1,
    "unit": 15,
    "reference": 4,
    "course_outcomes": 6,
    "experiment": 20,
    "course_metadata": 2,
    "project": 1
}

for chunk_type, expected in expected_counts.items():

    actual = chunk_counts.get(chunk_type, 0)

    status = "OK" if actual == expected else "CHECK"

    print(
        f"{chunk_type:<20}: "
        f"Actual = {actual:<3} "
        f"Expected = {expected:<3} "
        f"[{status}]"
    )


# ============================================================================
# 17. CHECK ALL UNIT CHUNKS
# ============================================================================

print("\n")
print("=" * 80)
print("UNIT CHUNKS")
print("=" * 80)

for i, chunk in enumerate(structure_chunks):

    if chunk.metadata.get("type") == "unit":

        print("\n" + "-" * 80)

        print("Chunk:", i)

        print(
            "Semester:",
            chunk.metadata.get("semester")
        )

        print(
            "Paper:",
            chunk.metadata.get("paper")
        )

        print(
            "Unit:",
            chunk.metadata.get("unit_number")
        )

        print(
            "Title:",
            chunk.metadata.get("unit_title")
        )

        print(
            "Lectures:",
            chunk.metadata.get("lectures")
        )

        print(
            "Pages:",
            chunk.metadata.get("start_page"),
            "to",
            chunk.metadata.get("end_page")
        )

        print("\nCONTENT:")
        print(chunk.page_content[:1200])


# ============================================================================
# 18. CHECK ALL EXPERIMENT METADATA
# ============================================================================

print("\n")
print("=" * 80)
print("ALL EXPERIMENT CHUNK METADATA")
print("=" * 80)

for chunk in structure_chunks:

    if chunk.metadata.get("type") == "experiment":

        print(
            "Experiment:",
            repr(chunk.metadata.get("experiment_number")),
            "| Paper:",
            repr(chunk.metadata.get("paper")),
            "| Semester:",
            repr(chunk.metadata.get("semester"))
        )


# ============================================================================
# 19. CHECK PRACTICAL PAPER-I EXPERIMENT 9
# ============================================================================

print("\n")
print("=" * 80)
print("CHECK PRACTICAL PAPER-I EXPERIMENT 9")
print("=" * 80)

found_experiment_9 = False

for chunk in structure_chunks:

    if (
        chunk.metadata.get("type") == "experiment"
        and str(
            chunk.metadata.get("experiment_number")
        ).strip() == "9"
        and (
            chunk.metadata.get("paper") is None
            or
            "Practical Paper - I"
            in str(chunk.metadata.get("paper"))
        )
    ):

        found_experiment_9 = True

        print("FOUND EXPERIMENT 9")

        print(
            "Page:",
            chunk.metadata.get("page",
            chunk.metadata.get("start_page"))
        )

        print(
            "Metadata:",
            chunk.metadata
        )

        print("\nCONTENT:")
        print(chunk.page_content)


if not found_experiment_9:

    print("Experiment 9 was NOT found.")


# ============================================================================
# 20. FINAL EXPERIMENT COUNT
# ============================================================================

experiment_chunks_final = [
    chunk
    for chunk in structure_chunks
    if chunk.metadata.get("type") == "experiment"
]

print("\n")
print("=" * 80)
print("FINAL EXPERIMENT COUNT")
print("=" * 80)

print(
    "Total experiment chunks:",
    len(experiment_chunks_final)
)

print(
    "Expected:",
    20
)

if len(experiment_chunks_final) == 20:
    print("STATUS: SUCCESS")
else:
    print("STATUS: CHECK")
```

```python
# ================================================================
# PAGE-LEVEL RETRIEVAL EVALUATION
# Statistics_syllabus.pdf
# Using 1-based PDF page numbers
# ================================================================

# ----------------------------------------------------------------
# 1. EVALUATION QUESTIONS
# ----------------------------------------------------------------

EVAL_QUERIES = [

    {
        "query": "What courses are included in Semester III?",
        "expected_page": 2
    },

    {
        "query": "What is Unit 5 of Probability Distributions-I?",
        "expected_page": 5
    },

    {
        "query": "What are the reference books for Probability Distributions-I?",
        "expected_page": 6
    },

    {
        "query": "What is Unit 3 of Probability Distributions-II?",
        "expected_page": 8
    },

    {
        "query": "What is the title of Experiment 9 in Practical Paper-I?",
        "expected_page": 10
    },

    {
        "query": "What is Unit 3 of Statistical Methods?",
        "expected_page": 12
    },

    {
        "query": "What is Unit 1 of Sampling Distributions and Exact Tests?",
        "expected_page": 14
    },

    {
        "query": "What is Experiment 9 in Practical Paper-II?",
        "expected_page": 16
    },

    {
        "query": "What are the project requirements in the syllabus?",
        "expected_page": 17
    },

    {
        "query": "What is the title of the B.Sc. Statistics program?",
        "expected_page": 1
    }
]


# ----------------------------------------------------------------
# 2. FUNCTION TO GET 1-BASED PAGE NUMBER
# ----------------------------------------------------------------

def get_page_number(doc):

    # ------------------------------------------------------------
    # Structure-aware chunks that already contain "page"
    # ------------------------------------------------------------

    if doc.metadata.get("page") is not None:

        page = doc.metadata["page"]

        return int(page)


    # ------------------------------------------------------------
    # Chunks containing start_page
    # ------------------------------------------------------------

    if doc.metadata.get("start_page") is not None:

        page = doc.metadata["start_page"]

        return int(page)


    # ------------------------------------------------------------
    # If no page information exists
    # ------------------------------------------------------------

    return None


# ----------------------------------------------------------------
# 3. CHECK PAGE METADATA
# ----------------------------------------------------------------

print("=" * 80)
print("PAGE METADATA CHECK")
print("=" * 80)

for i, doc in enumerate(structure_chunks[:10], start=1):

    print(f"\nChunk {i}")

    print("Type:",
          doc.metadata.get("type"))

    print("Page:",
          get_page_number(doc))

    print("Metadata:",
          doc.metadata)


# ----------------------------------------------------------------
# 4. CREATE STRUCTURE-AWARE RETRIEVER
# ----------------------------------------------------------------

structure_retriever = vector_store.as_retriever(
    search_kwargs={"k": 5}
)


# ----------------------------------------------------------------
# 5. EVALUATE RETRIEVER
# ----------------------------------------------------------------

def evaluate_retriever(
    retriever,
    eval_queries,
    k=5
):

    results = []


    for item in eval_queries:

        query = item["query"]

        expected_page = item["expected_page"]


        # --------------------------------------------------------
        # Retrieve top-k documents
        # --------------------------------------------------------

        retrieved_docs = retriever.invoke(query)

        retrieved_docs = retrieved_docs[:k]


        # --------------------------------------------------------
        # Get 1-based page numbers
        # --------------------------------------------------------

        retrieved_pages = [

            get_page_number(doc)

            for doc in retrieved_docs

        ]


        # --------------------------------------------------------
        # Check whether expected page appears
        # --------------------------------------------------------

        hit = expected_page in retrieved_pages


        # --------------------------------------------------------
        # Find rank of expected page
        # --------------------------------------------------------

        if hit:

            rank = retrieved_pages.index(
                expected_page
            ) + 1

        else:

            rank = None


        # --------------------------------------------------------
        # Save result
        # --------------------------------------------------------

        results.append({

            "query": query,

            "expected_page": expected_page,

            "retrieved_pages": retrieved_pages,

            "hit": hit,

            "rank": rank,

            "retrieved_docs": retrieved_docs
        })


    return results


# ----------------------------------------------------------------
# 6. CALCULATE HIT RATE AND MRR
# ----------------------------------------------------------------

def compute_metrics(results, k=5):

    n = len(results)


    if n == 0:

        print("No queries evaluated.")

        return


    # ------------------------------------------------------------
    # Hit Rate @ K
    # ------------------------------------------------------------

    hits = sum(

        1

        for r in results

        if r["hit"]

    )


    hit_rate = hits / n


    # ------------------------------------------------------------
    # Reciprocal Rank
    # ------------------------------------------------------------

    reciprocal_ranks = [

        1 / r["rank"]

        if r["hit"]

        else 0

        for r in results

    ]


    # ------------------------------------------------------------
    # Mean Reciprocal Rank
    # ------------------------------------------------------------

    mrr = sum(
        reciprocal_ranks
    ) / n


    # ------------------------------------------------------------
    # Print overall metrics
    # ------------------------------------------------------------

    print("\n" + "=" * 80)
    print("PAGE-LEVEL RETRIEVAL EVALUATION")
    print("=" * 80)

    print(
        f"\nNumber of queries : {n}"
    )

    print(
        f"Successful hits   : {hits}"
    )

    print(
        f"Hit Rate@{k}      : {hit_rate:.2%}"
    )

    print(
        f"MRR               : {mrr:.3f}"
    )


    # ------------------------------------------------------------
    # Print individual results
    # ------------------------------------------------------------

    print("\n" + "=" * 80)
    print("INDIVIDUAL QUERY RESULTS")
    print("=" * 80)


    for r in results:

        if r["hit"]:

            status = (
                f"HIT @ rank {r['rank']}"
            )

        else:

            status = "MISS"


        print(
            f"\n[{status}]"
        )

        print(
            "Query:"
        )

        print(
            r["query"]
        )

        print(
            "Expected page:",
            r["expected_page"]
        )

        print(
            "Retrieved pages:",
            r["retrieved_pages"]
        )


# ----------------------------------------------------------------
# 7. RUN THE EVALUATION
# ----------------------------------------------------------------

print("\nRunning page-level retrieval evaluation...\n")


results = evaluate_retriever(

    structure_retriever,

    EVAL_QUERIES,

    k=5

)


compute_metrics(

    results,

    k=5

)
```
